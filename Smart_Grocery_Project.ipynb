# Hybrid Product Recommendation System

```python
recommendation_data = (
    orders[["order_id", "customer_id"]]
    .merge(
        transactions[["order_id", "product_id", "quantity"]],
        on="order_id",
        how="inner"
    )
    .merge(
        products[["product_id", "product_name", "category"]],
        on="product_id",
        how="inner"
    )
)

recommendation_data.head()
```

```python
customer_product = recommendation_data.pivot_table(
    index="customer_id",
    columns="product_id",
    values="quantity",
    aggfunc="sum",
    fill_value=0
)

customer_product.head()
```

## 1. Collaborative Filtering

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
customer_similarity = cosine_similarity(customer_product)

similarity_df = pd.DataFrame(
    customer_similarity,
    index=customer_product.index,
    columns=customer_product.index
)
```

```python
customer_id = customer_product.index[0]

print("Customer ID:", customer_id)
```

## 2. Popularity Component

```python
product_popularity = (
    transactions
    .groupby("product_id")["quantity"]
    .sum()
    .sort_values(ascending=False)
)

product_popularity.head(10)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()

popularity_score = pd.Series(
    scaler.fit_transform(
        product_popularity.values.reshape(-1, 1)
    ).flatten(),
    index=product_popularity.index,
    name="popularity_score"
)

popularity_score.head(10)
```

```python
popularity_df = (
    popularity_score
    .reset_index()
    .merge(
        products[["product_id", "product_name", "category"]],
        on="product_id",
        how="left"
    )
    .sort_values(
        "popularity_score",
        ascending=False
    )
)

popularity_df.head(10)
```

```python
popularity_df.head(10)
```

## 3. Category Preference Component

```python
customer_category = (
    transactions
    .merge(
        orders[["order_id", "customer_id"]],
        on="order_id",
        how="left"
    )
    .merge(
        products[["product_id", "category"]],
        on="product_id",
        how="left"
    )
    .groupby(["customer_id", "category"])["quantity"]
    .sum()
)
```

```python
transactions.groupby("product_id")["quantity"].sum()
```

```python
category_preference = (
    customer_category
    .groupby(level=0)
    .transform(lambda x: x / x.max())
)

category_preference.name = "category_score"
```

```python
category_preference_df = (
    category_preference
    .reset_index()
    .sort_values(
        ["customer_id", "category_score"],
        ascending=[True, False]
    )
)

category_preference_df.head(10)
```

## 4. Hybrid Recommendation Model

```python
def recommend_products(customer_id, top_n=5):

    # Check customer
    if customer_id not in customer_product.index:
        return "Customer not found"

    # -----------------------------
    # 1. COLLABORATIVE SCORE
    # -----------------------------

    similar_customers = (
        similarity_df[customer_id]
        .drop(customer_id)
        .sort_values(ascending=False)
        .head(10)
    )

    collaborative_score = customer_product.loc[
        similar_customers.index
    ].T.dot(similar_customers)

    # Remove products already purchased
    purchased_products = customer_product.loc[customer_id]
    purchased_products = purchased_products[
        purchased_products > 0
    ].index

    collaborative_score = collaborative_score.drop(
        purchased_products,
        errors="ignore"
    )

    # Normalize collaborative score
    if collaborative_score.max() > 0:
        collaborative_score = (
            collaborative_score / collaborative_score.max()
        )


    # -----------------------------
    # 2. POPULARITY SCORE
    # -----------------------------

    popularity = popularity_score.reindex(
        collaborative_score.index
    ).fillna(0)


    # -----------------------------
    # 3. CATEGORY PREFERENCE
    # -----------------------------

    # Get category preference for this customer
    customer_categories = (
        category_preference_df[
            category_preference_df["customer_id"] == customer_id
        ]
        .set_index("category")["category_score"]
    )

    # Get category of each recommended product
    product_categories = products[
        ["product_id", "category"]
    ].set_index("product_id")

    category_score = (
        product_categories
        .loc[collaborative_score.index, "category"]
        .map(customer_categories)
        .fillna(0)
    )


    # -----------------------------
    # 4. HYBRID SCORE
    # -----------------------------

    hybrid_score = (
        0.6 * collaborative_score
        + 0.2 * popularity
        + 0.2 * category_score
    )

    # -----------------------------
    # 5. TOP RECOMMENDATIONS
    # -----------------------------

    recommendations = (
        hybrid_score
        .sort_values(ascending=False)
        .head(top_n)
        .reset_index()
    )

    recommendations.columns = [
        "product_id",
        "hybrid_score"
    ]

    # Add product details
    recommendations = recommendations.merge(
        products[
            ["product_id", "product_name", "category"]
        ],
        on="product_id",
        how="left"
    )

    return recommendations
```

```python
import random

random.seed(42)

customers = customer_product.index.tolist()

# Use up to 500 customers
test_customers = random.sample(
    customers,
    min(500, len(customers))
)

hits = 0
valid_tests = 0

for customer_id in test_customers:

    # Products purchased by customer
    purchased = customer_product.loc[customer_id]
    purchased = purchased[purchased > 0].index.tolist()

    # Need at least 2 products:
    # one to hide + at least one remaining purchase
    if len(purchased) < 2:
        continue

    # Hide one randomly selected product
    test_product = random.choice(purchased)

    # Create test copy
    customer_product_test = customer_product.copy()

    # Hide test product
    customer_product_test.loc[
        customer_id,
        test_product
    ] = 0

    # Generate recommendations
    recommendations = recommend_products_test(
        customer_id,
        5
    )

    # Check whether hidden product is recommended
    recommended_products = recommendations[
        "product_id"
    ].values

    if test_product in recommended_products:
        hits += 1

    valid_tests += 1


hit_rate_5 = hits / valid_tests

print("Valid test customers:", valid_tests)
print("Hits@5:", hits)
print("Hit Rate@5:", round(hit_rate_5, 4))
print("Hit Rate@5 (%):", round(hit_rate_5 * 100, 2), "%")
```

```python
import random

random.seed(42)

customers = customer_product.index.tolist()

test_customers = random.sample(
    customers,
    min(500, len(customers))
)

total_precision = 0
valid_tests = 0

for customer_id in test_customers:

    # Customer's purchased products
    purchased = customer_product.loc[customer_id]
    purchased = purchased[purchased > 0].index.tolist()

    # Need at least 2 products
    if len(purchased) < 2:
        continue

    # Hide one purchased product
    test_product = random.choice(purchased)

    # Create test copy
    customer_product_test = customer_product.copy()

    # Hide the test product
    customer_product_test.loc[
        customer_id,
        test_product
    ] = 0

    # Generate Top 5 recommendations
    recommendations = recommend_products_test(
        customer_id,
        5
    )

    recommended_products = set(
        recommendations["product_id"]
    )

    # Relevant product = hidden product
    relevant_products = {test_product}

    # Number of relevant recommendations
    relevant_count = len(
        recommended_products & relevant_products
    )

    # Precision@5 for this customer
    precision = relevant_count / 5

    total_precision += precision
    valid_tests += 1


precision_at_5 = total_precision / valid_tests

print("Valid test customers:", valid_tests)
print("Precision@5:", round(precision_at_5, 4))
print(
    "Precision@5 (%):",
    round(precision_at_5 * 100, 2),
    "%"
)
```
